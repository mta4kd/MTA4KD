1.1. The expected value of a single die roll is 3.5. This is based on taking the average of all possible outcomes: 

$$
\frac{1 + 2 + 3 + 4 + 5 + 6}{6} = 3.5.
$$

1.2. The expected value of the sum of two dice rolls is 7, assuming independence. This is because:

$$
E(X + Y) = E(X) + E(Y) = 3.5 + 3.5 = 7.
$$

1.3 In European roulette, there are 36 slots for the gambler to bet on, but **37 possible outcomes** (the extra slot being the green 0).  
This creates an inherent disadvantage to the gambler. Normally, the expected value of a bet would be $ \frac{1}{36} $ scaled by the wager,  
but in European roulette it becomes $ \frac{1}{37} $.

For a \$1 bet on a single color (red or black), the expected value per bet is:

$$
E[X] = \frac{18}{37}(+1) + \frac{19}{37}(-1) = \frac{18 - 19}{37} = -\frac{1}{37}.
$$

Thus, for every \$1 bet, the player loses on average:

$$
\boxed{E[X] = -0.027} \quad \text{(equivalently, a 2.7\% (-\$0.027) expected loss per spin).}
$$

1.4 In the scenario where rolling a 6 results in an additional roll, the expected value increases to 4.2 because of the following: 

$$
E[X] = \frac{1}{6}(1 + 2 + 3 + 4 + 5) + \frac{1}{6}(6 + E[X])
$$

$$
E[X] = 2.5 + 1 + \frac{E[X]}{6}
$$

$$
E[X] - \frac{E[X]}{6} = 3.5
$$

$$
\frac{5E[X]}{6} = 3.5
$$

$$
E[X] = 4.2
$$

$$
\boxed{E[X] = 4.2}
$$


1.5 Continuing the process until failing to roll a 6 results in a similar value (4.2) as seen by simulating the process.

```python
import random
import numpy as np

def play_game():
    total = 0
    while True:
        roll = random.randint(1, 6)
        if roll == 6:
            total += 6
        else:
            total += roll
            return total

# simulate many times
N = 1_000_000
results = [play_game() for _ in range(N)]

np.mean(results)
```

3.2 Show that ${V}(a + bX) = b^2 {V}(X)$:


$$
\begin{aligned}
\mathbb{V}[a + bX]
&= \int (a + bX - \mathbb{E}[a + bX])^2 f(x)\,dx \\
&= \int (a + bX - a - b\mathbb{E}[X])^2 f(x)\,dx \\
&= \int (bX - b\mathbb{E}[X])^2 f(x)\,dx \\
&= \int b^2 (X - \mathbb{E}[X])^2 f(x)\,dx \\
&= b^2 \int (X - \mathbb{E}[X])^2 f(x)\,dx \\
&= b^2 \mathbb{V}[X]
\end{aligned}
$$

3.3 Show that if $X$ is a normally distributed random variable, then $a + bX$ is distributed normally with mean $a+ b \mathbb{E}[X]$ and variance $b^2 \sigma_X^2$ :

$$
X \sim \mathcal{N}(\mu, \sigma^2)
$$

Let
$$
Y = a + bX.
$$

Compute the mean:

$$
\mathbb{E}[Y] = \mathbb{E}[a + bX]
= a + b\,\mathbb{E}[X]
= a + b\mu.
$$

Compute the variance:

$$
\mathbb{V}[Y] = \mathbb{V}[a + bX]
= b^2\,\mathbb{V}[X]
= b^2 \sigma^2.
$$

To show normality, write the density of \(Y\):

$$
f_Y(y)
= f_X\!\left(\frac{y-a}{b}\right)
\frac{1}{|b|}
= \frac{1}{|b| \sqrt{2\pi}\sigma}
\exp\!\left[
-\frac{1}{2}
\left(
\frac{ \frac{y-a}{b} - \mu }{\sigma }
\right)^2
\right].
$$

Rewrite:

$$
f_Y(y)
= \frac{1}{\sqrt{2\pi}\, |b|\sigma}
\exp\!\left[
-\frac{1}{2}
\left(
\frac{ y - (a + b\mu) }{\, b\sigma \,}
\right)^2
\right].
$$

Thus

$$
Y \sim \mathcal{N}(a + b\mu,\; b^2\sigma^2).
$$

4.1 Show that if $f_{XY}(x,y) = f_X(x) f_Y(y)$, then $\operatorname{cov}(X,Y) = 0$.

$$
\begin{aligned}
\operatorname{cov}(X,Y)
&= \int_{y} \int_{x} (x-\mathbb{E}[X])(y-\mathbb{E}[Y]) f_{XY}(x,y)\,dx\,dy \\
&= \int_{y} \int_{x} (x-\mathbb{E}[X])(y-\mathbb{E}[Y]) f_X(x) f_Y(y)\,dx\,dy \\
&= \int_{y} (y-\mathbb{E}[Y]) f_Y(y)
   \left[ \int_{x} (x-\mathbb{E}[X]) f_X(x)\,dx \right] dy \\
&= \int_{y} (y-\mathbb{E}[Y]) f_Y(y) \cdot 0 \, dy \\
&= 0
\end{aligned}
$$

4.2 An example (computation/simulation is fine) where $\text{cov}(X,Y)\approx 0$ but $f_{XY}(x,y)\neq 0$:

```python
from scipy.stats import gaussian_kde

# dependent but uncorrelated example
n = 200_000
X = np.random.uniform(-1, 1, size=n)
Y = X**2

# 1. covariance is approximately zero
cov_xy = np.cov(X, Y, bias=True)[0, 1]

# 2. joint density vs product of marginals (they should differ)
kde_joint = gaussian_kde(np.vstack([X, Y]))
kde_x = gaussian_kde(X)
kde_y = gaussian_kde(Y)

# pick a point to compare densities
x0, y0 = X[0], Y[0]

joint_density = kde_joint([x0, y0])
product_density = kde_x(x0) * kde_y(y0)

cov_xy, joint_density, product_density
```

Using the Taylor expansions,

$$
F(x+h) = F(x) + h f(x) + \frac{h^2}{2} f'(x) + O(h^3),
$$

and

$$
F(x-h) = F(x) - h f(x) + \frac{h^2}{2} f'(x) + O(h^3),
$$

subtract:

$$
F(x+h) - F(x-h)
= 2h f(x) + O(h^3).
$$

Then

$$
\mathbb{E}[\hat{f}_{X,h}(x)]
= \frac{F(x+h) - F(x-h)}{2h}
= f(x) + O(h^2).
$$
